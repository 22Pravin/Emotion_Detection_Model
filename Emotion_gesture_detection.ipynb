```python
from deepface import DeepFace
import matplotlib.pyplot as plt
import numpy as np

# Load your image path
img1 = "/content/passport photo.jpg"

# Extract face using the updated function
detected_faces = DeepFace.extract_faces(img_path=img1, enforce_detection=True)

# Check if a face was detected
if len(detected_faces) > 0:
    # Extract the first face (DeepFace returns a list of dictionaries)
    face = detected_faces[0]['face']

    # Display the detected face
    plt.imshow(face)
    plt.axis('off')  # Hide axes for better visualization
    plt.show()
else:
    print("No face detected in the image.")
```

```python
outcome=DeepFace.analyze(img1,actions=['emotion','age','gender'])
```
